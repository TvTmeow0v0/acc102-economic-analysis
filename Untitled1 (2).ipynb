# A Python-Based Comparison of GDP Growth, Inflation, and Unemployment in China, the United States, Japan, and India

## Problem Definition
This project compares GDP growth, inflation, and unemployment across four countries: China, the United States, Japan, and India. The aim is to identify differences in economic performance and stability.

## Intended User
The intended users are business students and general readers who want a simple overview of economic conditions in selected countries.

## Data Source
The dataset is based on country-level macroeconomic indicators collected from World Bank Open Data and organised into a CSV file for analysis.

## Access Date
Please replace this line with your actual data access date.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load the Dataset
In this section, the dataset is loaded and the first few rows are displayed.

```python
df = pd.read_csv("economic_data.csv")
df.head()
```

## Understand the Dataset
This section checks the structure, column names, missing values, and basic information of the dataset.

```python
df.info()
```

```python
df.columns
```

```python
df.isnull().sum()
```

## Data Cleaning
The dataset is checked for missing values and converted into suitable formats if necessary.

```python
df = df.dropna()
df["Year"] = df["Year"].astype(int)
df.head()
```

## Analysis 1: GDP Growth Comparison
This chart compares GDP growth across the four countries over time.

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x="Year", y="GDP_growth", hue="Country", marker="o")
plt.title("GDP Growth Comparison by Country")
plt.ylabel("GDP Growth (%)")
plt.xlabel("Year")
plt.grid(True)
plt.show()
```

## Analysis 2: Inflation Comparison
This chart compares inflation rates across the four countries over time.

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x="Year", y="Inflation", hue="Country", marker="o")
plt.title("Inflation Comparison by Country")
plt.ylabel("Inflation (%)")
plt.xlabel("Year")
plt.grid(True)
plt.show()
```

## Analysis 3: Unemployment Comparison
This chart compares unemployment rates across the four countries over time.

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x="Year", y="Unemployment", hue="Country", marker="o")
plt.title("Unemployment Comparison by Country")
plt.ylabel("Unemployment (%)")
plt.xlabel("Year")
plt.grid(True)
plt.show()
```

## Analysis 4: Average Indicator Comparison
This section compares the average GDP growth, inflation, and unemployment of each country over the selected period.

```python
average_data = df.groupby("Country")[["GDP_growth", "Inflation", "Unemployment"]].mean()
average_data
```

```python
average_data.plot(kind="bar", figsize=(10, 6))
plt.title("Average Economic Indicators by Country (2019–2023)")
plt.ylabel("Average Value")
plt.xlabel("Country")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

## Key Findings

1. India and China showed relatively stronger GDP growth than Japan and the United States over most of the selected years.
2. The United States experienced a sharp rise in inflation in 2022 compared with the other countries.
3. Japan had the lowest inflation rates in several years, suggesting a more stable price level.
4. India and the United States showed higher unemployment fluctuations during the selected period.
5. Economic conditions differed significantly across countries, showing that growth, inflation, and unemployment do not move in the same way.
